# Typy danych w Python

```python
W Python istnieje kilka predefiniowanych typów danych:
- liczbowe (int, float, complex)
- sekwencyjne typy danych (str, list, tuple)
- pozostałe typy (dict, set, frozenset)
```

## Typy liczbowe (Numeric Types)

```python
# Deklaracja zmiennej typu int
x,y = 42, -10


# Operacje arytmetyczne
suma = x + y  # 32
roznica = x - y  # 52
potega = x ** 2  # 1764
dzielenie_calkowite = x // 5  # 8

print(f"Suma: {suma}, Różnica: {roznica}, Potęga: {potega}, Dzielenie całkowite: {dzielenie_calkowite}")
```

```python
x,y = y,x
print(x,y)
```

```python
pow(x,2)
```

```python
# Deklaracja zmiennej typu float
pi = 3.14159
cena = 19.99

# Operacje arytmetyczne
pole_kola = pi * 5 ** 2  # 78.53975
cena_po_rabacie = cena * 0.9  # 17.991

print(f"Pole koła: {pole_kola:.4f}, Cena po rabacie: {cena_po_rabacie:.2f}")
```

```python
# Deklaracja zmiennej typu complex
z1 = 3 + 4j
z2 = 1 - 2j

# Operacje na liczbach zespolonych
suma_zespolona = z1 + z2  # (4+2j)
mnozenie_zespolone = z1 * z2  # (11-2j)

print(f"Suma zespolona: {suma_zespolona}, Mnożenie: {mnozenie_zespolone}")
```

```python
# Deklaracja zmiennej typu str
tekst = "Witaj, Python!"

# Operacje na stringach
dlugosc = len(tekst)  # 13
pierwsza_litera = tekst[0]  # 'W'
fragment = tekst[7:12]  # 'Python'
polaczenie = tekst + " 3"  # 'Witaj, Python! 3'

print(f"Długość: {dlugosc}, Pierwsza litera: {pierwsza_litera}, Fragment: {fragment}, Połączenie: {polaczenie}")
```

```python
imie = "Anna"
wiek = 25
saldo = 1234.5678
tekst = f"{imie:<15} ma {wiek} lat i saldo: {saldo:.2f} zł"
print(tekst)  # "      Anna ma 25 lat i saldo: 1234.57 zł"
```

```python
szablon = "Użytkownik {imie} ({id:04d}) zalogowany {czas}."
tekst = szablon.format(imie="Bartek", id=42, czas="2025-07-27")
print(tekst)  # Użytkownik Bartek (0042) zalogowany 2025-07-27.
```

```python
from string import Template
szablon = Template("Witaj, $imie! Masz $przedmiot.")
tekst = szablon.substitute(imie="Celina", przedmiot="książkę")
print(tekst)  # Witaj, Celina! Masz książkę.
```

```python
tekst = "zażółć"
zakodowany = tekst.encode('ascii', errors='ignore')  # Ignoruje znaki spoza ASCII
print(zakodowany)  # b'za'
```

```python
tekst = "Python 3.11"
print(tekst.startswith("Py"))  # True
print(tekst.endswith("11"))  # True
print(tekst.isalpha())  # False (zawiera cyfry i kropkę)
print(tekst.isalnum())  # False (zawiera kropkę)
```

```python
tekst = "   Python jest SUPER!   "
print(tekst.strip())  # "Python jest SUPER!"
print(tekst.lower())  # "   python jest super!   "
print(tekst.replace("SUPER", "świetny"))  # "   Python jest świetny!   "
```

```python
import unicodedata

tekst = "zażółć gęślą jaźń"
bez_diakrytykow = ''.join(c for c in unicodedata.normalize('NFD', tekst)
                          if unicodedata.category(c) != 'Mn')
print(bez_diakrytykow)  # zazolc gesla jazn
```

```python
import string
small_letters = string.ascii_lowercase
capital_letters = string.ascii_uppercase

for i,x in enumerate(zip(small_letters,capital_letters)):
    print(str(i+1)+". "+x[0]+" -> "+x[1])
```

```python
# Sortowanie według długości stringa, a potem alfabetycznie
slowa = ["python", "java", "c", "ruby"]
posortowane = sorted(slowa, key=lambda x: (len(x), x))
print(posortowane)  # ['c', 'java', 'ruby', 'python']
```

```python
liczby = [-2, -1, 0, 1, 2]
filter(lambda x: x > 0, liczby)
```

```python
liczby = [-2, -1, 0, 1, 2]
dodatnie = list(filter(lambda x: x**2 > 0, liczby))
print(dodatnie)  # [1, 2]
```

```python
# Deklaracja zmiennej typu list
lista = [1, "dwa", 3.0, True]

# Operacje na liście
lista.append(4)  # Dodaje 4 na końcu
lista[1] = "trzy"  # Zmienia element na indeksie 1
usuniety = lista.pop(2)  # Usuwa element na indeksie 2 (3.0)

print(f"Lista po zmianach: {lista}, Usunięty element: {usuniety}")
# Wynik: Lista po zmianach: [1, 'trzy', True, 4], Usunięty element: 3.0
```

```python
# Deklaracja zmiennej typu tuple
krotka = (10, "dwadzieścia", 30.5)

# Operacje na krotce
element = krotka[1]  # 'dwadzieścia'
dlugosc = len(krotka)  # 3
krotka_polaczona = krotka + (40,)  # (10, 'dwadzieścia', 30.5, 40)

print(f"Element: {element}, Długość: {dlugosc}, Połączona krotka: {krotka_polaczona}")
```
