```python
transakcje = [
    {"id": 1, "kwota": 100, "kategoria": "żywność"},
    {"id": 2, "kwota": 50, "kategoria": "transport"},
    {"id": 3, "kwota": 200, "kategoria": "żywność"},
    {"id": 4, "kwota": 75, "kategoria": "transport"}
]

# 1. Filtrowanie transakcji powyżej 50
duze_transakcje = [t for t in transakcje if t["kwota"] > 50]
print("Duże transakcje:", duze_transakcje)

# 2. Sumowanie kwot w kategorii "żywność"
suma_zywnosc = sum(t["kwota"] for t in transakcje if t["kategoria"] == "żywność")
print("Suma wydatków na żywność:", suma_zywnosc)

# 3. Grupowanie według kategorii
from itertools import groupby
from operator import itemgetter
posortowane = sorted(transakcje, key=lambda x: x["kategoria"])
grupowane = {k: list(g) for k, g in groupby(posortowane, key=lambda x: x["kategoria"])}
print("Grupowane według kategorii:", grupowane)
```
