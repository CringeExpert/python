```python
rzym = { 1000:"M", 900:"CM", 500:"D", 400:"CD", 100:"C", 90:"XC", 50:"L",
40:"XL", 10:"X", 9:"IX", 5:"V", 4:"IV", 1:"I" }

def liczby_rzymckie():
    x = int(input("Podaj liczbę całkowitą:"))
    print("Liczba",str(x),"w notacji rzymskiej to:",end=" ")
    r = list(rzym.keys())
    r.sort()
    r.reverse()
    lr = ""
    for i in r:
        while i <= x:
            lr += rzym[i]
            x -= i
    print(lr)
```

```python
liczby_rzymckie()
```

```python
r = list(rzym.keys())
```

```python
r.sort()
```

```python
print(r)
```

```python
r.reverse()
print(r)
```

```python
# Sortowanie listy słowników według wartości klucza
studenci = [
    {"imie": "Anna", "ocena": 4.5},
    {"imie": "Bartek", "ocena": 3.8},
    {"imie": "Celina", "ocena": 5.0}
]
# Sortowanie według oceny
posortowani = sorted(studenci, key=lambda x: x["ocena"], reverse=True)
print(posortowani)
# [{'imie': 'Celina', 'ocena': 5.0}, {'imie': 'Anna', 'ocena': 4.5}, {'imie': 'Bartek', 'ocena': 3.8}]
```

```python
W języku programowania Python istnieją cztery typy danych kolekcji:

-Lista to kolekcja, która jest uporządkowana i zmienna.
Dopuszcza duplikaty elementów.
-Krotka to kolekcja, która jest uporządkowana i niezmienna.
Dopuszcza duplikaty elementów.
-Set to kolekcja, która jest nieuporządkowana, niezmienna i nieindeksowana.
Brak duplikatów elementów.
-Słownik to kolekcja, która jest uporządkowana i zmienna.
Brak duplikatów elementów.
```
